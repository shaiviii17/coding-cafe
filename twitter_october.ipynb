```python
import json
from pathlib import Path
import pandas as pd

def load_ytd(path):
    raw = Path(path).read_text(encoding="utf-8")
    start = raw.find("[")
    if start == -1:
        raise ValueError("Could not find JSON data in the file.")
    data = json.loads(raw[start:])
    return pd.json_normalize(data)

tweets = load_ytd(tweets_path)

print("Tweets loaded:", len(tweets))
print("Column names:", tweets.columns.tolist())
```

```python

# Parse the timestamps as UTC
tweets["created_at_utc"] = pd.to_datetime(
    tweets["tweet.created_at"], utc=True, errors="coerce"
)

# Select tweets posted during October 2019
october = tweets[
    (tweets["created_at_utc"] >= "2019-10-01") &
    (tweets["created_at_utc"] < "2019-11-01")
].copy()

# 1. Total tweets in October
print("1. Total tweets:", len(october))

# 2. Retweets whose text starts with RT @
retweets = october["tweet.full_text"].fillna("").str.startswith("RT @")
print("2. Retweets:", int(retweets.sum()))

# 3. Day(s) with the most tweets
daily_counts = october.groupby(
    october["created_at_utc"].dt.day
).size()

print("3. Busiest day(s):", daily_counts[
    daily_counts == daily_counts.max()
].index.tolist())
print("Tweets on busiest day:", daily_counts.max())
```

```python

print("Valid timestamps:", tweets["created_at_utc"].notna().sum())

if not october.empty:
    print("Earliest October tweet:", october["created_at_utc"].min())
    print("Latest October tweet:", october["created_at_utc"].max())

print("October tweet count:", len(october))
print("Retweet count:", int(
    october["tweet.full_text"].fillna("").str.startswith("RT @").sum()
))
```

```python
import json
from pathlib import Path

target_id = "837881019105755136"

# Find the three files in your extracted archive
base = Path.home() / "Downloads" / "twitter-archive-class 2" / "data"

def load_js(path):
    text = path.read_text(encoding="utf-8")
    start = text.find("[")
    if start == -1:
        start = text.find("{")
    return json.loads(text[start:])

dm = load_js(base / "direct-messages.js")
following = load_js(base / "following.js")
followers = load_js(base / "follower.js")

# Inspect the direct-message structure without printing message text
print("DM data type:", type(dm).__name__)
print("DM top-level keys:", list(dm[0].keys()) if isinstance(dm, list) and dm else list(dm.keys()) if isinstance(dm, dict) else "none")
print("Following structure:", type(following).__name__)
print("Followers structure:", type(followers).__name__)
```

```python
# Find the conversation with the target user
target_id = "837881019105755136"

matches = []
for item in dm:
    conv = item.get("dmConversation", {})
    participants = conv.get("conversationId", "")
    if target_id in str(participants):
        matches.append(conv)

print("Matching conversations:", len(matches))

if matches:
    conv = matches[0]
    print("Conversation keys:", list(conv.keys()))
    messages = conv.get("messages", [])
    print("Messages in conversation:", len(messages))
    if messages:
        print("Message keys:", list(messages[0].keys()))
        print("Sender field names only — no message text shown.")
```

```python
# Count sent messages and check follow relationships
target_id = "837881019105755136"

messages = matches[0].get("messages", [])

# Inspect sender metadata only
for msg in messages:
    data = msg.get("messageCreate", {})
    print("Message fields:", list(data.keys()))

print("\nFollowing entry keys:",
      list(following[0].keys()) if following else [])
print("Follower entry keys:",
      list(followers[0].keys()) if followers else [])
```

```python
target_id = "837881019105755136"

# Question 2: count messages sent by the account owner.
# The owner is the sender who is NOT the target user.
messages = [m["messageCreate"] for m in matches[0]["messages"]]

owner_ids = {
    str(m["senderId"])
    for m in messages
    if str(m["senderId"]) != target_id
}

if len(owner_ids) == 1:
    owner_id = owner_ids.pop()
    sent_count = sum(str(m["senderId"]) == owner_id for m in messages)
    print("Q1 — Total messages:", len(messages))
    print("Q2 — Sent by account owner:", sent_count)

    # Extract user IDs from following/follower data
    def extract_ids(data, key):
        ids = set()
        for item in data:
            entry = item.get(key, {})
            if isinstance(entry, dict):
                value = entry.get("accountId") or entry.get("userId") or entry.get("id")
                if value is not None:
                    ids.add(str(value))
        return ids

    following_ids = extract_ids(following, "following")
    follower_ids = extract_ids(followers, "follower")

    follows_target = target_id in following_ids
    target_follows_owner = target_id in follower_ids

    print("Q3 — Mutual follow:", "Yes" if follows_target and target_follows_owner else "No")
else:
    print("Could not identify the account owner from message senders.")
    print("Sender IDs (IDs only):", [
        m["senderId"] for m in messages
    ])
```
